# Bell State Preparation

This notebook builds a two-qubit Bell state, executes a measurement-based simulation, and visualizes the observed measurement distribution.


```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
```

```python
qc = QuantumCircuit(2)
# Apply a Hadamard gate to create superposition
qc.h(0)
# Entangle the two qubits with a CNOT gate
qc.cx(0, 1)
# Apply a Z gate to introduce a phase flip
qc.z(0)
qc.measure_all()


qc.draw('mpl')
```

```python
sim = AerSimulator()

# Execute the circuit on the simulator with 2048 shots
result = sim.run(qc, shots=2048).result()

# Count how many times 00 and 11 are measured
counts = result.get_counts()

plot_histogram(counts)
```
