```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
```

```python
qc = QuantumCircuit(2)
#Hadamard
qc.h(0)
#CNOT
qc.cx(0, 1)
#Z
qc.z(0)
qc.measure_all()


qc.draw('mpl')
```

```python
sim = AerSimulator()

# Exécution du circuit sur le simulateur avec 2048 mesures
result = sim.run(qc, shots=2048).result()

#combien de fois 00 et '11' ont été mesurés
counts = result.get_counts()

plot_histogram(counts)
```
